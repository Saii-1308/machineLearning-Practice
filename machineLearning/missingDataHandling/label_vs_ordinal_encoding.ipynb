```python
import pandas as pd
import numpy as np
#when we don't have to assign rank we use label encoding

#Label Encoding 
#Label Encoding assigns a unique integer to each category without implying any order.
#It only converts categorical text data to numeric, but no meaning is attached to the numbers.
```

```python
df=pd.DataFrame(
    {'color':['red','green','blue','green','red','blue','red','green']}
)
df
```

```python
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder
lbl_encode=LabelEncoder()
```

```python
lbl_encode.fit_transform(df['color'])
```

```python
lbl_encode.transform(['red'])
```

```python
#when we have to assign rank we use ordinal encoding
#Ordinal Encoding is used when the categories have a meaningful order or ranking.
#The encoded numbers reflect the order of the categories.
from sklearn.preprocessing import OrdinalEncoder

df=pd.DataFrame(
    {
        'size':["small","medium","large","medium","small","large","small","medium"]
    }
)
df
```

```python
#create instance of OrdinalEncoder and then fit_transform

encoder=OrdinalEncoder(categories=[['small','medium','large']])

encoder.fit_transform(df[['size']])
```

```python
encoder.transform([['small']])
```

```python
#Target Guild Ordinal Encoding
#In Target Guided Ordinal Encoding, we replace each category with the mean of the target variable for that category.
#This method is particularly useful when there is a strong relationship between the categorical feature and the target
```

```python
df=pd.DataFrame({
    'city':['new york','londan','korea','japan','new york','korea','londan','japan'],
    'price':[500,200,300,400,450,350,250,420]
})
```

```python
df
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded']=df['city'].map(mean_price)
```

```python
df
```

```python
df[['price','city_encoded']]
```
